```python
import pandas as pd
import numpy as np
import re
import pickle
```

```python
df = pd.read_csv(r"C:\Users\Abdul\OneDrive\Documents\Ultimate AI Mastery BootCamp\NLP\Session 28,29&30\emotion_dataset.csv")
df.head()
```

```python
df["Emotion"].value_counts()
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text
```

```python
df["Text"] = df["Text"].apply(clean_text)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df["Emotion"] = encoder.fit_transform(df["Emotion"])
```

```python
X = df["Text"]
y = df["Emotion"]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=10000)
X = cv.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=5)
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
def predict_emotion(text):
    text = clean_text(text)
    text_vector = cv.transform([text])
    prediction = model.predict(text_vector)[0]
    return encoder.inverse_transform([prediction])[0]
```

```python
predict_emotion("I am feeling extremely happy today")
```

```python
predict_emotion("I am terrified about tomorrow")
```

```python
pickle.dump(model, open("emotion_model.pkl", "wb"))
pickle.dump(cv, open("vectorizer.pkl", "wb"))
pickle.dump(encoder, open("label_encoder.pkl", "wb"))
```
